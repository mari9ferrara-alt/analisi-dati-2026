# Analisi statistica degli asteroidi che si avvicinano alla Terra

## Obiettivo
Studiare i passaggi ravvicinati alla Terra degli asteroidi tra il 2000 e il 2025, mettendo in relazione distanza minima, velocità relativa e dimensione stimata. I dati provengono dall'API ufficiale NASA/JPL SBDB Close-Approach Data.

Un passaggio ravvicinato non implica automaticamente un rischio di impatto.

## 1. Importazione delle librerie

```python
import json
from urllib.parse import urlencode
from urllib.request import urlopen

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

plt.rcParams['figure.figsize'] = (9, 5)
plt.rcParams['axes.grid'] = True
```

## 2. Download dei dati NASA
Consideriamo soltanto gli asteroidi passati entro 0,05 unità astronomiche dalla Terra. Richiediamo anche il diametro misurato, quando disponibile, e il nome completo dell'oggetto.

```python
base_url = 'https://ssd-api.jpl.nasa.gov/cad.api'
parametri = {
    'date-min': '2000-01-01',
    'date-max': '2025-12-31',
    'dist-max': '0.05',
    'kind': 'a',
    'diameter': 'true',
    'fullname': 'true'
}

url = base_url + '?' + urlencode(parametri)
with urlopen(url, timeout=120) as risposta:
    dati_json = json.load(risposta)

print('Fonte:', dati_json['signature']['source'])
print('Versione API:', dati_json['signature']['version'])
print('Numero di passaggi scaricati:', dati_json['count'])
```

## 3. Costruzione e pulizia del DataFrame

```python
df = pd.DataFrame(dati_json['data'], columns=dati_json['fields'])

colonne_numeriche = [
    'jd', 'dist', 'dist_min', 'dist_max', 'v_rel', 'v_inf',
    'h', 'diameter', 'diameter_sigma'
]
for colonna in colonne_numeriche:
    df[colonna] = pd.to_numeric(df[colonna], errors='coerce')

df['data'] = pd.to_datetime(df['cd'], format='%Y-%b-%d %H:%M')
df['anno'] = df['data'].dt.year
df['nome'] = df['fullname'].str.strip()

AU_KM = 149_597_870.7
DISTANZA_LUNARE_KM = 384_400
df['distanza_km'] = df['dist'] * AU_KM
df['distanza_lunare'] = df['distanza_km'] / DISTANZA_LUNARE_KM
df['diametro_misurato_m'] = df['diameter'] * 1000

# Stima uniforme del diametro a partire dalla magnitudine assoluta H.
# Si assume un'albedo geometrica p = 0,14. Il risultato è indicativo.
ALBEDO_ASSUNTA = 0.14
df['diametro_stimato_m'] = (1329 / np.sqrt(ALBEDO_ASSUNTA)) * 10**(-df['h'] / 5) * 1000
df['entro_distanza_lunare'] = df['distanza_lunare'] < 1

df = df.sort_values('data').reset_index(drop=True)
df[['nome', 'data', 'distanza_lunare', 'v_rel', 'h', 'diametro_stimato_m']].head()
```

### Nota sulla dimensione
Il diametro reale è noto soltanto per una parte degli oggetti. Per confrontare l'intero campione calcoliamo quindi un diametro indicativo dalla magnitudine assoluta $H$, assumendo la stessa albedo per tutti gli asteroidi. Questa è un'approssimazione: oggetti con superfici diverse possono avere dimensioni reali differenti a parità di luminosità.

```python
print('Righe e colonne:', df.shape)
print('Asteroidi distinti:', df['des'].nunique())
print('Diametri misurati disponibili:', df['diameter'].notna().sum())
print('Valori mancanti principali:')
display(df[['dist', 'v_rel', 'h', 'diameter']].isna().sum())
```

## 4. Statistiche descrittive

```python
variabili = ['distanza_lunare', 'v_rel', 'diametro_stimato_m']
statistiche = df[variabili].describe().T
statistiche = statistiche.rename(columns={
    'count': 'numero', 'mean': 'media', 'std': 'deviazione standard',
    'min': 'minimo', 'max': 'massimo'
})
statistiche
```

## 5. Distribuzioni delle variabili

```python
fig, assi = plt.subplots(1, 3, figsize=(17, 4.5))

assi[0].hist(df['distanza_lunare'].dropna(), bins=40, color='royalblue', edgecolor='black')
assi[0].set_xlabel('Distanza minima [distanze lunari]')
assi[0].set_ylabel('Numero di passaggi')
assi[0].set_title('Distribuzione delle distanze')

assi[1].hist(df['v_rel'].dropna(), bins=40, color='darkorange', edgecolor='black')
assi[1].set_xlabel('Velocità relativa [km/s]')
assi[1].set_ylabel('Numero di passaggi')
assi[1].set_title('Distribuzione delle velocità')

assi[2].hist(df['diametro_stimato_m'].dropna(), bins=np.logspace(0, 4, 45), color='seagreen', edgecolor='black')
assi[2].set_xscale('log')
assi[2].set_xlabel('Diametro stimato [m]')
assi[2].set_ylabel('Numero di passaggi')
assi[2].set_title('Distribuzione delle dimensioni')

plt.tight_layout()
plt.show()
```

## 6. Numero di passaggi osservati per anno
Un aumento non implica necessariamente che nello spazio vi siano più asteroidi: può riflettere il miglioramento dei programmi di osservazione e catalogazione.

```python
passaggi_annui = df.groupby('anno').size()

plt.figure(figsize=(11, 5))
plt.plot(passaggi_annui.index, passaggi_annui.values, 'o-', color='purple')
plt.xlabel('Anno')
plt.ylabel('Numero di passaggi catalogati')
plt.title('Passaggi ravvicinati per anno')
plt.show()
```

## 7. Relazioni tra le variabili

```python
fig, assi = plt.subplots(1, 2, figsize=(15, 5))

assi[0].scatter(df['distanza_lunare'], df['v_rel'], s=8, alpha=0.25)
assi[0].set_xlabel('Distanza minima [distanze lunari]')
assi[0].set_ylabel('Velocità relativa [km/s]')
assi[0].set_title('Velocità e distanza')

assi[1].scatter(df['diametro_stimato_m'], df['distanza_lunare'], s=8, alpha=0.25, color='crimson')
assi[1].set_xscale('log')
assi[1].set_xlabel('Diametro stimato [m]')
assi[1].set_ylabel('Distanza minima [distanze lunari]')
assi[1].set_title('Dimensione stimata e distanza')

plt.tight_layout()
plt.show()
```

```python
dati_correlazione = df[['distanza_lunare', 'v_rel', 'h', 'diametro_stimato_m']].copy()
dati_correlazione['log10_diametro'] = np.log10(dati_correlazione['diametro_stimato_m'])
matrice_correlazione = dati_correlazione.drop(columns='diametro_stimato_m').corr()

fig, ax = plt.subplots(figsize=(8, 6))
immagine = ax.imshow(matrice_correlazione, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(matrice_correlazione.columns)))
ax.set_yticks(range(len(matrice_correlazione.columns)))
ax.set_xticklabels(matrice_correlazione.columns, rotation=35, ha='right')
ax.set_yticklabels(matrice_correlazione.columns)
for i in range(len(matrice_correlazione.columns)):
    for j in range(len(matrice_correlazione.columns)):
        ax.text(j, i, f'{matrice_correlazione.iloc[i, j]:.2f}', ha='center', va='center')
fig.colorbar(immagine, ax=ax, label='Coefficiente di correlazione')
ax.set_title('Matrice di correlazione')
plt.tight_layout()
plt.show()
```

La forte anticorrelazione tra $H$ e il diametro stimato è inevitabile, perché il diametro è stato calcolato direttamente da $H$: non costituisce quindi una scoperta indipendente.

## 8. Regressione lineare tra distanza e velocità

```python
campione_fit = df[['distanza_lunare', 'v_rel']].dropna()
risultato_fit = stats.linregress(campione_fit['distanza_lunare'], campione_fit['v_rel'])

x_fit = np.linspace(campione_fit['distanza_lunare'].min(), campione_fit['distanza_lunare'].max(), 200)
y_fit = risultato_fit.intercept + risultato_fit.slope * x_fit

plt.figure(figsize=(9, 5))
plt.scatter(campione_fit['distanza_lunare'], campione_fit['v_rel'], s=8, alpha=0.20, label='Dati')
plt.plot(x_fit, y_fit, color='red', linewidth=2, label='Regressione lineare')
plt.xlabel('Distanza minima [distanze lunari]')
plt.ylabel('Velocità relativa [km/s]')
plt.title('Regressione lineare')
plt.legend()
plt.show()

print(f'Pendenza = {risultato_fit.slope:.4f} km/s per distanza lunare')
print(f'Intercetta = {risultato_fit.intercept:.4f} km/s')
print(f'Coefficiente r = {risultato_fit.rvalue:.4f}')
print(f'R^2 = {risultato_fit.rvalue**2:.4f}')
print(f'p-value della pendenza = {risultato_fit.pvalue:.3e}')
```

```python
residui = campione_fit['v_rel'] - (risultato_fit.intercept + risultato_fit.slope * campione_fit['distanza_lunare'])

fig, assi = plt.subplots(1, 2, figsize=(14, 4.5))
assi[0].scatter(campione_fit['distanza_lunare'], residui, s=8, alpha=0.25)
assi[0].axhline(0, color='red')
assi[0].set_xlabel('Distanza minima [distanze lunari]')
assi[0].set_ylabel('Residuo [km/s]')
assi[0].set_title('Residui del modello lineare')

assi[1].hist(residui, bins=45, edgecolor='black')
assi[1].set_xlabel('Residuo [km/s]')
assi[1].set_ylabel('Frequenza')
assi[1].set_title('Distribuzione dei residui')

plt.tight_layout()
plt.show()
```

Un $p$-value piccolo non basta a dimostrare una relazione importante: con un campione molto grande anche effetti debolissimi possono risultare statisticamente significativi. Per questo bisogna osservare soprattutto $r$, $R^2$ e il grafico dei residui.

## 9. Passaggi entro la distanza della Luna

```python
n = len(df)
k = int(df['entro_distanza_lunare'].sum())
frequenza = k / n
alpha = 0.05
limite_inferiore = 0 if k == 0 else stats.beta.ppf(alpha / 2, k, n - k + 1)
limite_superiore = 1 if k == n else stats.beta.ppf(1 - alpha / 2, k + 1, n - k)

print('Passaggi entro una distanza lunare:', k)
print('Passaggi totali:', n)
print(f'Frequenza osservata: {100 * frequenza:.2f}%')
print(f'Intervallo di confidenza esatto al 95%: [{100*limite_inferiore:.2f}%, {100*limite_superiore:.2f}%]')
```

```python
confronto = df.groupby('entro_distanza_lunare').agg(
    numero=('des', 'size'),
    velocita_media=('v_rel', 'mean'),
    velocita_std=('v_rel', 'std'),
    diametro_mediano=('diametro_stimato_m', 'median')
)
confronto.index = ['Oltre la distanza lunare', 'Entro la distanza lunare']
confronto
```

## 10. Test di Kolmogorov-Smirnov
Confrontiamo la distribuzione delle velocità dei passaggi entro una distanza lunare con quella dei passaggi più lontani.

```python
velocita_vicini = df.loc[df['entro_distanza_lunare'], 'v_rel'].dropna()
velocita_lontani = df.loc[~df['entro_distanza_lunare'], 'v_rel'].dropna()
test_ks = stats.ks_2samp(velocita_vicini, velocita_lontani)

plt.figure(figsize=(9, 5))
plt.hist(velocita_lontani, bins=45, density=True, alpha=0.55, label='Oltre 1 distanza lunare')
plt.hist(velocita_vicini, bins=45, density=True, alpha=0.55, label='Entro 1 distanza lunare')
plt.xlabel('Velocità relativa [km/s]')
plt.ylabel('Densità di probabilità')
plt.title('Confronto delle distribuzioni delle velocità')
plt.legend()
plt.show()

print(f'Statistica KS = {test_ks.statistic:.4f}')
print(f'p-value = {test_ks.pvalue:.3e}')
```

## 11. Grafico tridimensionale

```python
campione_3d = df[['diametro_stimato_m', 'v_rel', 'distanza_lunare', 'anno']].dropna()
if len(campione_3d) > 4000:
    campione_3d = campione_3d.sample(4000, random_state=42)

fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111, projection='3d')
punti = ax.scatter(
    np.log10(campione_3d['diametro_stimato_m']),
    campione_3d['v_rel'],
    campione_3d['distanza_lunare'],
    c=campione_3d['anno'], cmap='viridis', s=10, alpha=0.55
)
ax.set_xlabel('log10(diametro stimato [m])')
ax.set_ylabel('Velocità relativa [km/s]')
ax.set_zlabel('Distanza minima [distanze lunari]')
ax.set_title('Dimensione, velocità e distanza degli asteroidi')
fig.colorbar(punti, ax=ax, pad=0.1, label='Anno')
plt.show()
```

## 12. I passaggi più ravvicinati
